```python
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
#Fibra
ref_fiber = np.loadtxt('../Ref_Fiber.txt')
sample_fiber = np.loadtxt('../Samp_Fiber.txt')
ddpm_fiber = ref_fiber - sample_fiber
np.savetxt('Fiber.txt', ddpm_fiber)
plt.figure(figsize=(10, 6))
plt.subplot(1, 3, 1)
plt.imshow(ref_fiber, cmap='gray')
plt.title('Referencia Fibra')
plt.subplot(1, 3, 2)
plt.imshow(sample_fiber, cmap='gray')
plt.title('Muestra Fibra')
plt.subplot(1, 3, 3)
plt.imshow(ddpm_fiber, cmap='gray')
plt.title('DDPM Fibra')
#Sphere
ref_sphere = np.loadtxt('../Ref_Sphere.txt')
sample_sphere = np.loadtxt('../Samp_Sphere.txt')
ddpm_sphere = ref_sphere - sample_sphere
np.savetxt('Sphere.txt', ddpm_sphere)
plt.figure(figsize=(10, 6))
plt.subplot(1, 3, 1)
plt.imshow(ref_sphere, cmap='gray')
plt.title('Referencia Esfera')
plt.subplot(1, 3, 2)
plt.imshow(sample_sphere, cmap='gray')
plt.title('Muestra Esfera')
plt.subplot(1, 3, 3)
plt.imshow(ddpm_sphere, cmap='gray')
plt.title('DDPM Esfera')
```
